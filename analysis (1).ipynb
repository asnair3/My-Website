### Part 0.0: Data Reading Functions

```python
from data_utils import read_csv_rows, get_keys, columnar, head, select, convert_columns_to_int, concat
data_rows_izzi: list[dict[str, str]] = read_csv_rows(DATA_FILE_NAME_IZZI)
data_rows_alyssa: list[dict[str, str]] = read_csv_rows(DATA_FILE_NAME_ALYSSA)
print(f"Loaded {len(data_rows_izzi)} rows from Izzi and {len(data_rows_alyssa)} rows from Alyssa.")
```

```python
izzi_cols: dict[str, list[str]] = columnar(data_rows_izzi)
alyssa_cols: dict[str, list[str]] = columnar(data_rows_alyssa)
combined_cols: dict[str, list[str]] = concat(izzi_cols, alyssa_cols)
print(f"Combined columns: {get_keys(combined_cols)}")
```

## Part 1: Analysis and Visualizations

In this section, we process the integrated dataset and create visualizations to explore student trends.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Select and Clean Data
relevant_cols = ["major", "difficulty", "hours_online_work", "understanding"]
data_subset = select(combined_cols, relevant_cols)
data_ready = convert_columns_to_int(data_subset, ["difficulty", "understanding"])

# 2. Major Distribution
plt.figure(figsize=(10,6))
major_ax = sns.countplot(data=data_ready, x="major")
major_ax.set_title("Distribution of Student Majors")
plt.xticks(rotation=45, ha='right')

# 3. Weekly Study Hours
hours_dist = sns.displot(data=data_ready, x="hours_online_work", aspect=1.5, color="teal")
hours_dist.set(title="Distribution of Weekly Study Hours", xlabel="Hours per Week")

# 4. Difficulty vs. Understanding
rel_plot = sns.relplot(data=data_ready, x="difficulty", y="understanding", hue="major")
rel_plot.set(title="Difficulty vs. Understanding")

# Save the three charts as separate PNG files for the website
# Note: 'major_ax' and 'rel_plot' are the variables you named your plots
major_ax.get_figure().savefig("major_distribution.png", bbox_inches='tight')
hours_dist.savefig("hours_histogram.png")
rel_plot.savefig("difficulty_scatter.png")
```
